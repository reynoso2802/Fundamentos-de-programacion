```python
print ("Calculadora de propina por persona")
total = float(input("Ingresa el total de la cuenta:"))
propina = int(input("ingresa el porcentaje de propina:")) 
personas = int(input("¿Cuantas personas son?"))

propina = (total * propina /100) / personas 

print ("$" + str(propina) + " por persona")
```
