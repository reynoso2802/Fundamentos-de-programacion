Ejercicio 14

```python
lista = [7, 3, 9, 2]

def maximo_lista(lista):

    mayor = lista[0]

    for numero in lista:

        if numero > mayor:
            mayor = numero

    return mayor


resultado = maximo_lista(lista)

print("El elemento máximo es:", resultado)
```

Ejercicio 18

```python
def factorial(num):

    if num == 0 or num == 1:
        return 1

    else:
        return num * factorial(num - 1)


num = int(input("Ingresa un número: "))

resultado = factorial(num)

print("El factorial de", num, "es:", resultado)
```

Ejercicio 21

```python
calificaciones = {
    "Ana": [8, 9, 10],
    "Beto": [7, 7, 8]
}

for nombre in calificaciones:

    total = 0

    for calificacion in calificaciones[nombre]:

        total = total + calificacion

    promedio = total / len(calificaciones[nombre])

    print(f"{nombre}: {promedio:.2f}")
```
