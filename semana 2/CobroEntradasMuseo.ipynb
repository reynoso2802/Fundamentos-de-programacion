```python
# Precios
p1 = 0
p2 = 30
p3 = 45
total = 0

cvisit = int(input("¿Cuantos visitantes ingresan?:"))

for i in range(cvisit):

    print("\nVisitante", i + 1)

    edad = int(input("Edad del visitante: "))
    tipo = str(input("Tipo de visitante: "))

    # Precio según edad
    if edad < 3:
        precio = p1
    elif edad <= 17:
        precio = p2
    else:
        precio = p3

    subtotal = precio

    # Descuentos
    if tipo == "adulto mayor":
        descuento = precio * 0.12
    elif tipo == "profesor" or tipo == "estudiante":
        descuento = precio * 0.10
    else:
        descuento = 0

    precio_final = precio - descuento

    total = total + precio_final

    # Desglose
    print("Subtotal: $" + str(subtotal))
    print("Descuento aplicado: $" + str(descuento))
    print("Precio final: $" + str(precio_final))

print("\nTotal general: $" + str(total))
```
