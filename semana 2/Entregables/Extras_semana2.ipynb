```python
total = 0

while True:
    precio = float(input("Precio del boleto:"))

    if precio == 0:
        print("Menor de 3 años sin cargo")
        continue
    total = total + precio
    if total >= 100:
        print("Aforo limite alcanzado, se detiene el registro")
        break

print(f"Total acumulado: ${total:.2f}")
```

```python
visit = int(input("Número de visitantes: "))

adultos = 0
sumaedades = 0

for i in range(visit):
    edad = int(input("Edad del visitante " + str(i + 1) + ": "))
    sumaedades = sumaedades + edad

    if edad >= 18:
        adultos = adultos + 1

prom = sumaedades / visit

print("Adultos:", adultos)
print(f"Promedio de edad: {prom:.2f}")
```
