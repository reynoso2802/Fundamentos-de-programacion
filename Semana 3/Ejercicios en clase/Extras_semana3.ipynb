Ejercicio 6:

```python
dividendo = int(input("Ingresa el dividendo: "))
divisor = int(input("Ingresa el divisor: "))

cociente = 0
residuo = dividendo

while residuo >= divisor:
    residuo = residuo - divisor
    cociente = cociente + 1

print(dividendo, "÷", divisor, "=", cociente, "con residuo", residuo)
```
