```python
matriz = [
    [1,2,3],
    [4,5,6],
    [7,8,9],
]
def impresion_matriz(matriz):
    for fila in matriz:
        for num in fila:
            print(num, end=" ")
    print("")

def suma_matriz(matriz):
    total = 0 
    for fila in matriz:
        for num in fila:
            total = total + num 

    return total 

impresion_matriz(matriz)
resultado = suma_matriz(matriz)
print ("La suma de todos los elementos de la matriz es:", resultado)


```

```python
matriz = [
    [1,2,3],
    [4,5,6],
    [7,8,9],
]

def suma_fila(fila):
    total = 0
    for num in fila:
        total = total + num 
    return total 

def mostrar_tabla(matriz):
    for fila in matriz:
        for num in fila:
            print(num, end=" ")
        print ("")

mostrar_tabla(matriz) 

for f in range (len(matriz)):
    resultado = suma_fila(matriz[f])
    print ("Suma de la fila", f, ":", resultado)
```

```python
matriz = [
    [1,2,3],
    [4,5,6],
    [7,8,9],
    [2,4,6],
]

def multiplicar_columna(matriz,columna):
    prod = 1
    for f in range (len(matriz)):
        num = matriz [f][columna]
        nuevo_prod = 0
        for i in range(num):
            nuevo_prod = nuevo_prod + prod
        prod = nuevo_prod
    return prod 

columna = int(input("numero de columna: "))

while columna < 0 or columna > 2:
    print ("Ingresa una columna entre 0 y 2")

    columna = int(input("NUmero de columna: "))

resultado =multiplicar_columna(matriz,columna)
print ("el producto de la columna", columna, "es:",resultado)
```
