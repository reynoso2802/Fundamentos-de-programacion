```python
#Función para sumar calificaciones

def sumar_calificaciones(calificaciones):

    suma = 0

    for cal in calificaciones:
        suma = suma + cal

    return suma


#Tupla de calificaciones

calif = (7.5, 9.0, 8.0, 6.5, 10.0)

print("Tercera calificación:", calif[2])

#Se piden las nuevas calificaciones al usuario

cal1 = float(input("Nueva calificación 1: "))
cal2 = float(input("Nueva calificación 2: "))

#Se crea la nueva tupla

calificaciones = calif + (cal1, cal2)

print("Nueva tupla:", calificaciones)

#Se convierte en lista y se ordena

lista = list(calificaciones)

lista.sort(reverse = True)

print("Lista ordenada (mayor a menor):", lista)

#Se obtiene la suma

total = sumar_calificaciones(calificaciones)

print("Suma total:", total)
```

```python
#Para buscar un contacto

def buscar_contacto(agenda, nombre):

    if nombre in agenda:
        return agenda[nombre]

    else:
        return None


#Diccionario con los contactos

agenda = {
    "Ana": "5598876354",
    "Luis": "3345672873",
    "Mía": "22235657653"
}

#Se piden los datos del nuevo contacto

nombre = input("Nombre del nuevo contacto: ")
telefono = input("Teléfono del nuevo contacto: ")

agenda[nombre] = telefono

#Se imprimen los contactos registrados

print("Contactos registrados:", ", ".join(agenda.keys()))

#Se pide el contacto que se quiere buscar

buscar = input("Nombre a buscar: ")

resultado = buscar_contacto(agenda, buscar)

if resultado != None:

    print("El teléfono de", buscar, "es:", resultado)

else:

    print("Contacto no encontrado")
```

```python
#Función para contar las palabras

def contar_palabras(mensaje):

    palabras = mensaje.split()

    return len(palabras)


#Mensaje inicial

mensaje = "Python es un lenguaje poderoso"

print("Longitud del mensaje:", len(mensaje))

print("En mayúsculas:", mensaje.upper())

nuevo_mensaje = mensaje.replace("Python", "programación")

print("Texto reemplazado:", nuevo_mensaje)

total = contar_palabras(mensaje)

print("Palabras totales:", total)
```

```python
#Tupla con la lista de espera

espera = ("María", "José", "Carlos", "Lucía", "Pedro")

print("Tercer elemento:", espera[2])

#Se solicitan los dos nombres nuevos

nombre1 = input("Ingresa un nuevo nombre: ")
nombre2 = input("Ingresa otro nombre: ")

#Se agregan a una nueva tupla

espera = espera + (nombre1, nombre2)

print("Nueva tupla:", espera)

#Se convierte en lista para ordenarla

lista = list(espera)

lista.sort()

print("Lista ordenada:", lista)

#Cantidad de personas

cant = len(lista)

print("Cantidad total de personas:", cant)
```

```python
candidatos = {
    "A" : 0,
    "B" : 0,
    "C" : 0
    }

cantidad = int(input("Cuantos votos serán?: "))
conteo = 1

while conteo <= cantidad: 
    voto = input(f"ingresa el voto {conteo} (A,B,C):")
    if voto.upper () in candidatos:
        print(f"Voto válido para {voto.upper()}")
        candidatos[voto.upper()] += 1
        conteo += 1
    else:
        print("El voto no es valido, vota nuevamente")

print("\nResultados:")
for k, v in candidatos.items():
    print (f"Opcion {k}, votos {v}")
```
